```python
# Fill missing values with the most frequent value (mode)
mod = df["fraud reported"].mode()[0]
df["fraud reported"] = df["fraud reported"].fillna(mod)
```

```python
# CSV file does not contain duplicate rows
print("Duplicate rows before removing:", df.duplicated().sum())
df = df.drop_duplicates()
print("Duplicate rows after removing:", df.duplicated().sum())
```

```python
print(df["claim_date"].head(10))
```

```python
#Fix Incorrect Data Types

# Convert claim_date to datetime
df['claim_date'] = pd.to_datetime(df['claim_date'], errors='coerce')

# Convert annual_income to numeric
df['annual_income'] = pd.to_numeric(df['annual_income'], errors='coerce')

# Convert age_of_vehicle to numeric
df['age_of_vehicle'] = pd.to_numeric(df['age_of_vehicle'], errors='coerce')

# Convert injury_claim to numeric
df['injury_claim'] = pd.to_numeric(df['injury_claim'], errors='coerce')

print(df.dtypes)
```

```python
#Clean Messy Column Names

df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(' ', '_', regex=False)
)

print(df.columns.tolist())
```

```python
#Encode Categorical Columns

categorical_cols = df.select_dtypes(include='object').columns

df_encoded = pd.get_dummies(
    df,
    columns=categorical_cols,
    drop_first=True
)

print("Encoded Shape:", df_encoded.shape)
```

```python
df_encoded.to_csv("cleaned_data.csv", index=False)
```

```python
#Scale Numeric Columns
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

numeric_cols = df_encoded.select_dtypes(include=['int64', 'float64']).columns

df_encoded[numeric_cols] = scaler.fit_transform(df_encoded[numeric_cols])

print(df_encoded.head())
```

```python
Q1 = df['total_claim'].quantile(0.25)
Q3 = df['total_claim'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df = df[(df['total_claim'] >= lower) & (df['total_claim'] <= upper)]

print("Outliers removed successfully!")
print("New Shape:", df.shape)
```

```python
import matplotlib.pyplot as plt

# Select only numeric columns
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns

for col in numeric_cols:
    # Calculate IQR
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    # Calculate outlier limits
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Find outliers
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]

    # Print information
    print("=" * 50)
    print(f"Column: {col}")
    print(f"Number of Outliers: {len(outliers)}")

    # Draw box plot
    plt.figure(figsize=(6,4))
    plt.boxplot(df[col].dropna())
    plt.title(f"Box Plot of {col}")
    plt.ylabel(col)
    plt.show()
```
